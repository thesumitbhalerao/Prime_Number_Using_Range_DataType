#### Wap to check given number is prime or not ?
##### input from user -> iterate upto which no you want to check
##### check how many factor of that no -> if factor are only 1 and itself then prime, if more than 2 factors then not prime
##### ex. n=7 -> range(1,n+1) -> range(1,8) -> 1,2,3,4,5,6,7 -> factors of 7 -> 1,7 -> count=2 -> it is prime number
##### ex. n=4 -> range(1,n+1) -> range(1,5) -> 1,2,3,4-> factors of 4 -> 1,2,4 -> count=3 -> not prime number

```python
n=int(input("Enter a number to check whether it is prime or not : "))
print("number is :",n)
count=0
for i in range(1,n+1):
    if n%i==0:
        count+=1
if count == 2:
    print("prime number")
else:
    print("not prime number")
```
